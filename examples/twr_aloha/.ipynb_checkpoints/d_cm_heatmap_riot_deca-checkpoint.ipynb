```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import statistics
```

```python
df = pd.read_csv('data.csv')
df2 = df[['d_err_absolute', 'src', 'nei']]
df2 = df2.assign(d_err_absolute_cm=lambda x: round(x.d_err_absolute * 100))
df3 = df2.pivot_table(index='src', columns='nei', values='d_err_absolute_cm', aggfunc = statistics.mean).fillna(0).astype(int)
sns.heatmap(data=df3, annot=True)
```
